# Practical Task 1 
Create a ChromaDB collection and store at least 10 sample text documents.

```python
import chromadb
```

```python


# Create a ChromaDB client
client = chromadb.PersistentClient(path="./chroma_db")

# Create or get a collection
collection = client.get_or_create_collection(
    name="sample_documents"
)

# 10 sample documents
documents = [
    "Artificial Intelligence enables machines to perform tasks that normally require human intelligence.",
    "Machine Learning is a branch of Artificial Intelligence that allows computers to learn from data.",
    "Deep Learning uses neural networks with multiple layers to learn complex patterns.",
    "Python is a popular programming language used in Artificial Intelligence and Machine Learning.",
    "Natural Language Processing helps computers understand and process human language.",
    "Computer Vision enables machines to understand and analyze images and videos.",
    "RAG combines information retrieval with Large Language Models to generate better answers.",
    "Embeddings represent text as numerical vectors that capture semantic meaning.",
    "ChromaDB is a vector database used to store and search embeddings efficiently.",
    "Similarity Search finds documents that are semantically similar to a user's query."
]

# Unique IDs for each document
ids = [f"doc_{i}" for i in range(1, 11)]

# Add documents to the collection
collection.add(
    ids=ids,
    documents=documents
)

print("10 documents successfully stored in ChromaDB.")

# Display the stored documents
results = collection.get()

for doc_id, document in zip(results["ids"], results["documents"]):
    print(f"{doc_id}: {document}")
```

# Practical Task 2 
Perform similarity searches for the following questions:<br>
• What is the internship fee?<br>
• How long is the internship? <br>
• Is placement assistance available? <br>
Observe the retrieved results.

```python


# Create ChromaDB client
client = chromadb.PersistentClient(path="./chroma_db")

# Get the collection created in Practical Task 1
collection = client.get_or_create_collection(
    name="internship_documents"
)

# Sample internship documents
documents = [
    "The internship fee is Rs. 5,000 for the complete internship program.",
    "The internship duration is 6 months.",
    "Placement assistance is available for eligible candidates after successful completion of the internship.",
    "The internship provides practical training in Artificial Intelligence and Machine Learning.",
    "Students receive hands-on experience through real-world projects.",
    "The program includes training in Python programming.",
    "Participants learn Machine Learning algorithms and techniques.",
    "The internship includes projects related to Deep Learning.",
    "Students receive guidance from experienced mentors throughout the program.",
    "A certificate is provided after successful completion of the internship."
]

ids = [f"internship_{i}" for i in range(1, 11)]

# Add documents to ChromaDB
collection.add(
    ids=ids,
    documents=documents
)

# Questions for similarity search
questions = [
    "What is the internship fee?",
    "How long is the internship?",
    "Is placement assistance available?"
]

# Perform similarity searches
for question in questions:
    print("\nQuestion:", question)

    results = collection.query(
        query_texts=[question],
        n_results=3
    )

    print("Retrieved Results:")

    for i, document in enumerate(results["documents"][0], start=1):
        print(f"{i}. {document}")
```

Include their advantages and common use cases.

| Similarity Measure     | How It Works                                                                                                      | Advantages                                                                           | Common Use Cases                                                         |
| ---------------------- | ----------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------ | ------------------------------------------------------------------------ |
| **Cosine Similarity**  | Measures the angle between two vectors, focusing on their direction.                                              | Works well with high-dimensional text embeddings; less affected by vector magnitude. | **Text similarity, semantic search, RAG, document retrieval**            |
| **Euclidean Distance** | Measures the straight-line distance between two vectors. Smaller distance means greater similarity.               | Simple to understand and effective when vector magnitude is meaningful.              | **Clustering, image similarity, recommendation systems**                 |
| **Dot Product**        | Multiplies corresponding vector values and adds the results. Higher score generally indicates greater similarity. | Fast to calculate and efficient for large-scale vector search.                       | **Neural networks, recommendation systems, vector databases, retrieval** |


# Practical Task 5 
Write a Python program that accepts a user query and displays the Top 3 most 
similar document chunks from ChromaDB.

```python


# Connect to ChromaDB
client = chromadb.PersistentClient(path="./chroma_db")

# Get the collection
collection = client.get_or_create_collection(
    name="internship_documents"
)

# Accept query from the user
query = input("Enter your question: ")

# Perform similarity search
results = collection.query(
    query_texts=[query],
    n_results=3
)

# Display results
print("\nTop 3 Most Similar Document Chunks:")
print("-" * 50)

for i, document in enumerate(results["documents"][0], start=1):
    print(f"\nResult {i}:")
    print(document)
```

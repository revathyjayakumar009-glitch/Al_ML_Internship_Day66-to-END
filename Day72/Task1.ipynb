# 1. Define Similarity Search. 

Similarity Search is a technique used to find documents, text, or data that are most similar in meaning or content to a given query. In RAG, text is converted into embeddings (vectors), and similarity search compares the query vector with stored vectors to retrieve the most relevant information.

# 2. Differentiate Keyword Search and Similarity Search. 

| Keyword Search                                                                       | Similarity Search                                                                               |
| ------------------------------------------------------------------------------------ | ----------------------------------------------------------------------------------------------- |
| Searches for exact words or phrases.                                                 | Searches for similar meaning or context.                                                        |
| Depends mainly on matching keywords.                                                 | Uses vector embeddings to compare meaning.                                                      |
| May miss relevant results with different wording.                                    | Can find relevant results even when wording is different.                                       |
| Example: Searching **"internship duration"** finds documents containing those words. | Searching **"How long is the internship?"** can find **"The internship duration is 6 months."** |
| Commonly used in traditional search engines.                                         | Commonly used in RAG and vector databases.                                                      |


# 3. Explain Cosine Similarity. 

Cosine Similarity is a method used to measure how similar two vectors are by calculating the cosine of the angle between them.
In RAG, text is converted into vectors (embeddings). Cosine similarity compares the query vector with document vectors to determine which text is most relevant.<br>

* 1 → Very similar<br>
* 0 → Not similar<br>
* -1 → Opposite direction<br>

For example, if the query “How long is the internship?” and a document containing “The internship duration is 6 months” have a high cosine similarity score, that document is likely to be retrieved.

# 4. What is Top-K Retrieval? 

Top-K Retrieval is a technique in which a system retrieves the K most relevant documents or text chunks from a collection based on their similarity to the user's query.

In RAG, after calculating similarity scores, the system selects the top K highest-scoring chunks and passes them to the LLM for generating the final answer.

Example:
If K = 3, the system retrieves the 3 most relevant chunks for the user's question.

In short: Top-K Retrieval means selecting the K most relevant results from a similarity search.

# 5. Why is Similarity Search essential in RAG? 

Similarity Search is essential in RAG because it helps the system find and retrieve the most relevant information from a large collection of documents.<br>

It is important because:<br>

* It finds information based on meaning, not just exact keywords.<br>
* It retrieves the most relevant text chunks for a user's question.<br>
* It provides the LLM with relevant context to generate accurate answers. <br>
* It reduces irrelevant information being sent to the LLM.<br>
* It improves the accuracy and reliability of RAG applications.<br>

In short: Similarity Search connects the user's question with the right information stored in the knowledge base.

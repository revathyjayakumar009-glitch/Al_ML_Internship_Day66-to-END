# 1. Define a PDF Question Answering Chatbot.  

A PDF Question Answering Chatbot is an AI application that allows users to ask questions about the content of a PDF document and receive relevant answers automatically. It uses techniques such as document loading, text chunking, embeddings, vector databases, retrieval, and Large Language Models (LLMs). In a RAG-based system, LangChain manages the workflow and ChromaDB stores and retrieves document embeddings.

# 2. Explain the complete workflow. 

The complete workflow of a PDF Question Answering Chatbot follows the RAG (Retrieval-Augmented Generation) approach:<br>

PDF Document → PDF Loader → Text Chunking → Embeddings → ChromaDB → Retriever → LLM → Final Answer<br>

1. Upload PDF: The user provides a PDF document to the chatbot.
2. PDF Loading: A PDF loader extracts the text from the document.
3. Text Chunking: The extracted text is divided into smaller, meaningful chunks.
4. Embedding Generation: Each chunk is converted into a numerical vector using an embedding model.
5. Store in ChromaDB: The vectors are stored in ChromaDB, a vector database.
6. User Question: The user asks a question about the PDF.
7. Retrieval: The question is converted into an embedding, and the retriever searches ChromaDB for the most relevant chunks.
8. LLM Processing: The retrieved information is provided as context to the LLM.
9. Generate Answer: The LLM generates a relevant answer based on the retrieved PDF content.
10. Display Answer: The chatbot displays the answer to the user.

# 3. What is LangChain?  

LangChain is an open-source framework used to build applications powered by Large Language Models (LLMs). It provides tools and components for connecting LLMs with external data sources, documents, APIs, and vector databases.

In a PDF Question Answering Chatbot, LangChain helps manage the complete RAG workflow, including loading PDFs, splitting text into chunks, generating embeddings, retrieving relevant information, and sending the retrieved context to an LLM.

In simple words: LangChain acts as a framework that connects different components together to build an intelligent AI application.

# 4. Why do we use ChromaDB?  

ChromaDB is a vector database used to store, manage, and search text embeddings efficiently.<br>

In a PDF Question Answering Chatbot, ChromaDB is used to:<br>

* Store embeddings of PDF text chunks.
* Quickly find text chunks that are most similar to a user's question.
* Support semantic search rather than only keyword matching.
* Provide relevant information to the LLM for generating accurate answers.
* Make the retrieval process faster and more efficient.

# 5. Explain similarity search. 

Similarity Search is a technique used to find the most relevant information by comparing the embedding/vector of a user's query with the embeddings stored in a vector database such as ChromaDB.<br>

In a PDF Question Answering Chatbot:<br>

* The user's question is converted into an embedding.<br>
* The embedding is compared with the stored PDF text embeddings.
* ChromaDB calculates how similar the vectors are.
* The most similar text chunks are retrieved.
* These relevant chunks are provided to the LLM to generate the answer.

#### Example:<br>
If the user asks, "How long is the internship?", similarity search finds the PDF section containing information about the internship duration, even if the exact words are different.

# **Practical Task 2**
Load a sample PDF using PyPDFLoader. 

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader("Day 71-RAG.pdf") 
documents = loader.load()
```

# **Practical Task 3** 
Split the PDF into chunks using RecursiveCharacterTextSplitter. 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Create text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

# Split the PDF documents into chunks
chunks = text_splitter.split_documents(documents)

# Display number of chunks
print("Number of chunks:", len(chunks))

# Display the first chunk
print("\nFirst Chunk:")
print(chunks[0].page_content)
```

# **Practical Task 4**
Store the chunks in ChromaDB and perform a similarity search for three different questions. 

```python
!pip install -U langchain-huggingface
```

```python
!pip install -U langchain-chroma
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
# Create embedding model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Store PDF chunks in ChromaDB
db = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="day71_rag"
)

print("PDF chunks stored successfully in ChromaDB.")
```

```python
questions = [
    "What is a PDF Question Answering Chatbot?",
    "Why do we use ChromaDB?",
    "What is the purpose of chunk overlap?"
]

for question in questions:
    print("\n" + "=" * 60)
    print("Question:", question)

    results = db.similarity_search(question, k=3)

    for i, doc in enumerate(results, 1):
        print(f"\nRelevant Chunk {i}:")
        print(doc.page_content)
```

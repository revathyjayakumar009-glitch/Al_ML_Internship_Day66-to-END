# 1. FAISS (Facebook AI Similarity Search)

FAISS is an open-source vector database library developed by Meta (Facebook AI). It is designed for efficient similarity search and clustering of dense vectors. FAISS is widely used in RAG applications to quickly retrieve the most relevant document embeddings.

### Key Features
1. Fast similarity search on large datasets.
2. Supports CPU and GPU acceleration.
3. Open-source and free to use.
4. Easy integration with LangChain.
5. Ideal for local RAG applications.
### Advantages
1. High search speed.
2. Works well with millions of vectors.
3. No internet connection required.
### Limitation
Primarily designed for local storage and does not provide cloud hosting.

# 2. ChromaDB

ChromaDB is an open-source vector database designed specifically for AI applications. It stores text embeddings and enables semantic search, making it a popular choice for RAG chatbots and document retrieval systems.

### Key Features
1. Easy to install and use.
2. Stores embeddings with document metadata.
3. Integrates with LangChain and LlamaIndex.
4. Suitable for small to medium-sized AI projects.
5. Supports persistent local storage.
### Advantages
1. Beginner-friendly.
2. Lightweight and open source.
3. Good for rapid prototyping.
### Limitation
Less scalable than enterprise cloud vector databases.

# 3. Pinecone

Pinecone is a managed cloud-based vector database that provides fast and scalable similarity search for AI applications. It is widely used in production RAG systems because it automatically manages infrastructure, indexing, and scaling.

### Key Features
1. Fully managed cloud service.
2. Highly scalable for millions or billions of vectors.
3. Real-time vector search.
4. Automatic indexing and maintenance.
5. Integrates with LangChain and other AI frameworks.
### Advantages
1. No server management required.
2. High performance and reliability.
3. Suitable for enterprise applications.
### Limitation
Requires an internet connection and may have usage costs.

Explain how a chatbot can answer questions from a PDF using the RAG approach.

A Retrieval-Augmented Generation (RAG) chatbot can answer questions from a PDF by retrieving relevant information from the document before generating a response.
### Steps in the RAG Process
#### 1. Load the PDF
The chatbot reads the PDF using a document loader.
#### 2. Split the Text
The PDF content is divided into smaller chunks for efficient processing.
#### 3.Create Embeddings
Each text chunk is converted into numerical vectors (embeddings).
#### 4.Store in a Vector Database
The embeddings are stored in a vector database such as FAISS, ChromaDB, or Pinecone.
#### 5. User Asks a Question
Example: "What is the admission fee?"
#### 7.Retrieve Relevant Information
* The question is converted into an embedding.
* The vector database searches for the most similar text chunks in the PDF.
#### 8. Generate the Answer
* The retrieved text and the user's question are sent to the Large Language Model (LLM).
* The LLM generates an accurate answer based on the PDF content.

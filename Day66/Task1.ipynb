# 1. Define RAG.  

Retrieval-Augmented Generation (RAG) is an AI technique that enhances a Large Language Model (LLM) by retrieving relevant information from external knowledge sources—such as documents, PDFs, databases, or websites—and using that information to generate accurate and context-aware responses.

### Simple Definition

RAG combines information retrieval with text generation. Instead of relying only on the model's built-in knowledge, it first searches for relevant information and then uses it to answer the user's question.

### Example
* User: "What is the BCA fee at ABC College?"
* RAG chatbot:
1. Searches the college.txt document.
2. Finds the line: "BCA Fee: ₹50,000".
3. Responds: "The BCA fee at ABC College is ₹50,000."

### Key Benefits of RAG:

1. Uses custom knowledge from your own documents.
2. Produces more accurate and up-to-date answers.
3. Reduces incorrect or made-up responses (hallucinations).
4. No need to retrain the AI model when documents are updated.

# 2. Explain the RAG workflow. 


The Retrieval-Augmented Generation (RAG) workflow combines information retrieval with a Large Language Model (LLM) to generate accurate answers using external knowledge.

### RAG Workflow Steps
#### 1. User asks a question
Example: "What is the BCA fee?"
#### 2.Convert the question into an embedding
The question is converted into a numerical vector (embedding) for similarity search.
#### 3.Search the vector database
The embedding is compared with stored document embeddings in a vector database (e.g., FAISS or ChromaDB).
#### 4.Retrieve relevant document chunks
The system retrieves the most relevant sections (chunks) from the documents.
#### 5.Build the prompt
The retrieved context is combined with the user's question.
#### 6.Send to the Large Language Model (LLM)
The LLM receives both the question and the retrieved context.
#### 7.Generate the final answer
The LLM produces an accurate and context-aware response based on the retrieved information.

# 4. What is an embedding?  

An embedding is a numerical (vector) representation of text, words, or sentences that captures their meaning. It converts text into a list of numbers so that AI models can compare and understand the similarity between different pieces of text.<br>
## Example

Text:<br>
Artificial Intelligence
Embedding (vector):<br>
[0.45, -0.12, 0.89, 0.33, ...]
### How Embeddings Work
1.Convert text into numerical vectors.
2.Store these vectors in a vector database.
3.Compare vectors to find the most similar information.
4.Retrieve the most relevant document for answering the user's question.
### Importance of Embeddings
1. Help AI understand the meaning of text.
2. Enable semantic (meaning-based) search.
3. Improve document retrieval in RAG systems.
4. Increase the accuracy of chatbot responses.
### Example in RAG
1. Document: "BCA fee is ₹50,000."
2. User Question: "How much does BCA cost?"
3. Even though the wording is different, embeddings recognize that both refer to the same meaning, allowing the chatbot to retrieve the correct information.

# 5. What is a vector database? 

A vector database is a specialized database that stores embeddings (numerical vectors) and enables fast similarity searches. It helps RAG systems quickly find the most relevant documents related to a user's question.
### How It Works
1. Documents are converted into embeddings (vectors).
2. The embeddings are stored in a vector database.
3. When a user asks a question, the question is also converted into an embedding.
4. The vector database compares the question embedding with the stored embeddings.
5. It retrieves the most similar document chunks, which are then sent to the LLM to generate the final answer.
### Example

Suppose the database contains these document chunks:<br>

* Document 1: "BCA fee is ₹50,000."
* Document 2: "The college is located in Kochi."
* Document 3: "B.Com course duration is 3 years."<br>

User Question: "How much is the BCA fee?"<br>

The vector database finds that Document 1 is the most similar and returns it to the chatbot.<br>

### Popular Vector Databases
* FAISS
* ChromaDB
* Pinecone
* Weaviate
* Milvus
### Advantages
1. Fast similarity search.
2. Efficient storage of embeddings.
3. Improves the accuracy of RAG chatbots.
4. Handles large collections of documents effectively.

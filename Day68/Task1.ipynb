# 1. Define an embedding. 

An embedding is a numerical representation of data such as text, images, or words in the form of a vector (a list of numbers).

In NLP, an embedding converts text into numbers while capturing its meaning and relationships.

### Example:
"Python is easy to learn"
        ↓
Embedding Model
        ↓
[0.21, -0.45, 0.73, 0.12, ...]### 

# 2. Explain semantic search. 

Semantic search is a search technique that finds information based on the meaning and context of a query, rather than only matching exact keywords.

It uses embeddings to convert the user's query and documents into numerical vectors and then compares their similarity.
## Example:

User searches:<br>
"What can I use Python for?"<br>
A semantic search system can find: <br>
"Python is widely used for data analysis, machine learning,<br>
web development, and automation."<br>
Even though the exact words "what can I use Python for" are not present, the meanings are similar.


# 3. What is a Sentence Transformer?  

A Sentence Transformer is a type of AI model that converts sentences or text into numerical vectors called embeddings while capturing their semantic meaning.

### For example:<br>
"I love learning Python." <br>
        ↓<br>
Sentence Transformer<br>
        ↓
[0.21, -0.43, 0.67, ...]<br>
These embeddings can then be used to compare sentence meanings, perform semantic search, document similarity, and build RAG systems.

# 4. Differentiate keyword search and semantic search.  

| Keyword Search                                                              | Semantic Search                                                             |
| --------------------------------------------------------------------------- | --------------------------------------------------------------------------- |
| Searches for **exact words or keywords**.                                   | Searches based on **meaning and context**.                                  |
| Mainly matches the words in the query with the document.                    | Compares the meaning of the query with the document.                        |
| May fail if different words are used for the same meaning.                  | Can find relevant results even when different words are used.               |
| Usually uses techniques like **TF-IDF** or inverted indexes.                | Uses **embeddings and vector similarity**.                                  |
| Faster and simpler to implement.                                            | More powerful for understanding natural language.                           |
| Example: Search **"car repair"** and find documents containing those words. | Search **"How can I fix my vehicle?"** and find documents about car repair. |

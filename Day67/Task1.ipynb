# 1. Explain the complete RAG pipeline.  

RAG (Retrieval-Augmented Generation) is a technique that combines information retrieval with Large Language Models (LLMs). It allows an AI system to retrieve relevant information from external documents, such as PDFs, and use that information to generate accurate answers.

# 5. Explain the role of a vector database.  

A vector database is a database used to store and search vector embeddings. In a RAG system, it stores the embeddings of document chunks and helps find the chunks that are most relevant to a user's question.

### Main Functions
1. Stores embeddings – Saves the numerical representations of document chunks.
2. Stores document information – Keeps the original text or metadata associated with each vector.
3. Performs similarity search – Finds document chunks that are semantically similar to the user's question.
4. Provides relevant context – Sends the retrieved information to the LLM for generating an answer.
5. Handles large amounts of data efficiently – Makes searching through many document chunks faster.
### Examples of Vector Databases
1. FAISS
2. ChromaDB
3. Pinecone
### Example

If a PDF contains information about college admissions and the user asks:<br>

"What documents are required for admission?"<br>

The vector database searches the stored embeddings and retrieves the chunks containing the relevant admission requirements.

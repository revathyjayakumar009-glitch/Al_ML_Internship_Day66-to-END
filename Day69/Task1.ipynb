# 1. Define a Vector Database.  

A Vector Database is a specialized database used to store, manage, and search vector embeddings.

It stores data as numerical vectors and allows us to find similar or related information using similarity search.

Example: In a RAG chatbot, document embeddings are stored in a vector database. When a user asks a question, the database searches for the most relevant document chunks and returns them to the AI model.

Examples: ChromaDB, FAISS, Pinecone, and Weaviate.

# 2. Explain ChromaDB.  

ChromaDB is an open-source vector database designed to store, manage, and search vector embeddings generated from text, documents, images, and other data.

It is commonly used in AI applications, semantic search, and RAG (Retrieval-Augmented Generation) systems.

### Key Features:
* Stores vector embeddings efficiently.
* Performs similarity searches.
* Supports metadata and document storage.
* Easy to use with Python.
* Useful for building RAG applications.<br>
Can store and retrieve relevant document chunks based on a user's query.

# 3. Explain FAISS.  

FAISS (Facebook AI Similarity Search) is an open-source library developed by Meta AI for storing and performing fast similarity searches on vector embeddings.

It is widely used in AI applications, semantic search, recommendation systems, and RAG systems.

### Key Features:
* Performs fast similarity searches.
* Works with large collections of vectors.
* Supports different indexing techniques.
* Efficient for finding the nearest or most similar vectors.
* Works well with Python and NumPy.<br>
Can be used with embedding models to build semantic search systems.

# 5. Why are vector databases used in RAG? 

Vector databases are used in RAG (Retrieval-Augmented Generation) to store and quickly retrieve relevant information using vector embeddings.

### Main Reasons:
1. Store embeddings – Document chunks are converted into numerical vectors and stored.
2. Semantic search – They find information based on meaning, not just exact keywords.
3. Fast retrieval – Quickly retrieve the most relevant document chunks for a user's question.
4. Improve AI answers – The retrieved information is provided to the LLM as context, helping it generate more accurate answers.
5. Handle large datasets – They can efficiently search through thousands or millions of embeddings.
